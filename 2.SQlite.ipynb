```python
#day2：SQLite、SQL基础，JION，目标：能够独立查询数据

from idlelib import query

import pandas as pd
import os
import glob

from pandas.core.interchange import dataframe

#读取数据
data_path = './data/'
#读取所有数据
csv_files = glob.glob(data_path + '*.csv')
#建立字典
dataframes = {}
#遍历数据
for file in csv_files:
    name = os.path.basename(file).replace('.csv', '')  #替换将后缀替换为空
    dataframes[name] = pd.read_csv(file)


orders= dataframes[('olist_orders_dataset')]
orders.head()
```

```python
import sqlite3
#创建数据库链接
conn = sqlite3.connect('olist.db')
#将所有dataframe写入数据库
for name,df in dataframes.items():
    df.to_sql(
        name,
        conn,
        if_exists='replace',
        index=False
    )
    print('成功')#9个数据9个成功
```

```python
#SQL查询语句
query="""
SELECT *
FROM olist_orders_dataset
LIMIT 10;
"""
result = pd.read_sql_query(query,conn)
result
```

```python
query="""
SELECT order_id,customer_id
FROM olist_orders_dataset
"""
result = pd.read_sql_query(query,conn)
result
```

```python
query="""
SELECT *
FROM olist_orders_dataset
WHERE order_status ='delivered'
LIMIT 10;
"""
result = pd.read_sql_query(query,conn)
result
```

```python
#ASC升序，早到晚，小到大。DESC降序晚到早，大到小
query="""
SELECT *
FROM olist_orders_dataset
ORDER BY order_purchase_timestamp ASC
LIMIT 10;
"""
result = pd.read_sql_query(query,conn)
result
```

```python
query="""
SELECT order_status,COUNT(*) AS order_count
FROM olist_orders_dataset
GROUP BY order_status;
"""
result = pd.read_sql_query(query,conn)
result
```

```python
query="""
SELECT *
FROM olist_orders_dataset
INNER JOIN olist_order_items_dataset
ON olist_orders_dataset.order_id=olist_order_items_dataset.order_id
LIMIT 10;
"""
result = pd.read_sql_query(query,conn)
result
```

```python
query = """
SELECT
    o.order_id,
    o.customer_id,
    o.order_status,
    i.product_id,
    i.price,
    i.freight_value
FROM olist_orders_dataset AS o
INNER JOIN olist_order_items_dataset AS i
ON o.order_id = i.order_id
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)
result
```
