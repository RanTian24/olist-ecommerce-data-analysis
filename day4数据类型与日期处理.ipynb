```python
#导入库
import pandas as pd
import numpy as np
import sqlite3
import os
import glob
#设置pandas显示选项
pd.set_option('display.max_columns', None)
pd.set_option("display.width",1000)
#设置文件路径

data_path = './data/'
#读取所有数据
csv_files = glob.glob(data_path + '*.csv')
#建立字典
dataframes = {}
#遍历数据
for file in csv_files:
    name = os.path.basename(file).replace('.csv', '')  #替换将后缀替换为空
    dataframes[name] = pd.read_csv(file)
```

```python
#查看订单表的数据类型
orders = dataframes["olist_orders_dataset"]
print(orders.dtypes)
```

```python
#转换日期格式
data_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]
for col in data_columns:
    orders[col] = pd.to_datetime(orders[col])#依次取出在in后名称的列名，存到变量col，再将order表中的col转换为时间格式

print(orders.dtypes)
```

```python
#计算订单配送时长

orders["delivery_days"] = (
    orders["order_delivered_customer_date"] - orders["order_purchase_timestamp"]
).dt.total_seconds() / 86400

orders[[
    "order_id",
    "order_purchase_timestamp",
    "order_delivered_customer_date",
    "delivery_days"
]].head(10)
```

```python
#统计配送时长
print("平均配送时长", orders["delivery_days"].mean())
print("最长配送时长", orders["delivery_days"].max())
print("最短配送时长", orders["delivery_days"].min())
#配送时长最长有209天，存在异常
```

```python
#找出最长的10笔订单
slow_orders = orders.sort_values(
    by = "delivery_days",
    ascending = False
)

slow_orders[[
    "order_id",
    "order_purchase_timestamp",
    "order_delivered_customer_date",
    "delivery_days"
]].head(10)
```

```python
#判断这些订单是否超过了预算日期。
orders["delay_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_estimated_delivery_date"]
).dt.total_seconds() / 86400#创建一个delay——day

slow_orders = orders.sort_values(
    by = "delivery_days",
    ascending = False
)#慢订单的定义

slow_orders[[
    "order_id",
    "delivery_days",
    "order_estimated_delivery_date",
    "order_delivered_customer_date",
    "delay_days"
]].head(10)
```

```python
#整体延迟情况
print("延迟订单数量",(orders["delay_days"]>0).sum())

print("提前订单数量",(orders["delay_days"]<=0).sum())

print("平均延时天数",orders.loc[orders["delay_days"]>0, "delay_days"].mean())
```
