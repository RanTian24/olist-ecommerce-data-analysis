```python
import pandas as pd
import os
import glob
data_path = "./data/"

csv_files = glob.glob(data_path + "*.csv")

dataframes = {}

for file in csv_files:
    name = os.path.basename(file).replace(".csv", "")
    dataframes[name] = pd.read_csv(file)

#查找主键有没有重复值
primary_keys={
    'olist_orders_dataset':['order_id'],
    'olist_customers_dataset':['customer_id'],
    "olist_sellers_dataset": ["seller_id"],
    'olist_products_dataset':['product_id'],
    'olist_order_items_dataset':['order_id','order_item_id'],
}
```

```python
for table ,keys in primary_keys.items():
    df =dataframes[table]
    duplicate_count=df.duplicated(subset=keys).sum()
    null_count=df[keys].isnull().sum()

print(f'\n数据表：{table}')
print(f'候选主键：{keys}')
print(f'重复数量：{duplicate_count}')
print(f'控制数量：{null_count}')
```

```python
orders = dataframes['olist_orders_dataset']
items = dataframes['olist_order_items_dataset']
```

```python
#检查订单号，能否在订单表中找到，将items从order中提取转换为集合，再减去orders中的集合，因为orders>=items，所有如果有数则证明异常。（外键关联）
missing_orders=(
    set(items['order_id'])-
    set(orders['order_id'])
)
print('无法关联到的订单数量：',len(missing_orders))
```
