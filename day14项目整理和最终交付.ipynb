````python
sql_content = """
-- =====================================
-- Olist 巴西电商数据分析项目
-- 第一部分：基础数据查询
-- =====================================

-- 1. 查询订单表全部数据
SELECT *
FROM orders;

-- 2. 查询已完成的订单
SELECT *
FROM orders
WHERE order_status = 'delivered';

-- 3. 查询订单状态为 canceled 的订单
SELECT *
FROM orders
WHERE order_status = 'canceled';

-- 4. 按订单创建时间降序排列
SELECT *
FROM orders
ORDER BY order_purchase_timestamp DESC;

-- 5. 查询已完成订单，并按下单时间从早到晚排列
SELECT *
FROM orders
WHERE order_status = 'delivered'
ORDER BY order_purchase_timestamp ASC;
"""

with open(sql_file, "a", encoding="utf-8") as file:
    file.write("\n" + sql_content)

print("基础数据查询语句写入完成")
````

````python
group_content = """
-- =====================================
-- 第二部分：GROUP BY 分组统计
-- =====================================

-- 1. 统计不同订单状态的订单数量
SELECT
    order_status,
    COUNT(order_id) AS order_count
FROM orders
GROUP BY order_status;

-- 2. 统计不同客户的订单数量
SELECT
    customer_id,
    COUNT(order_id) AS order_count
FROM orders
GROUP BY customer_id;

-- 3. 统计不同支付方式的使用次数
SELECT
    payment_type,
    COUNT(*) AS payment_count
FROM order_payments
GROUP BY payment_type;

-- 4. 统计不同支付方式的支付总金额
SELECT
    payment_type,
    SUM(payment_value) AS total_payment
FROM order_payments
GROUP BY payment_type
ORDER BY total_payment DESC;
"""

with open(sql_file, "a", encoding="utf-8") as file:
    file.write("\n" + group_content)

print("GROUP BY 分组统计语句写入完成")
````

````python
join_content = """
-- =====================================
-- 第三部分：JOIN 多表关联查询
-- =====================================

-- 1. 关联订单表与订单商品明细表
SELECT
    o.order_id,
    o.order_status,
    oi.product_id,
    oi.price
FROM orders AS o
JOIN order_items AS oi
    ON o.order_id = oi.order_id;

-- 2. 关联订单表与客户表
SELECT
    o.order_id,
    o.order_status,
    c.customer_unique_id,
    c.customer_city
FROM orders AS o
JOIN customers AS c
    ON o.customer_id = c.customer_id;

-- 3. 关联订单表与支付表
SELECT
    o.order_id,
    o.order_status,
    p.payment_type,
    p.payment_value
FROM orders AS o
JOIN order_payments AS p
    ON o.order_id = p.order_id;

-- 4. 统计每个订单的商品销售金额
SELECT
    order_id,
    SUM(price) AS sales_amount
FROM order_items
GROUP BY order_id
ORDER BY sales_amount DESC;
"""

with open(sql_file, "a", encoding="utf-8") as file:
    file.write("\n" + join_content)

print("JOIN 多表关联语句写入完成")
````

````python
summary_content = """
-- =====================================
-- 第四部分：SQL 综合分析
-- =====================================

-- 1. 统计订单总数
SELECT
    COUNT(order_id) AS total_orders
FROM orders;

-- 2. 统计不同订单状态的数量及占比
SELECT
    order_status,
    COUNT(*) AS order_count,
    ROUND(
        COUNT(*) * 100.0 / (SELECT COUNT(*) FROM orders),
        2
    ) AS order_percentage
FROM orders
GROUP BY order_status
ORDER BY order_count DESC;

-- 3. 统计商品总销售额及商品销量
SELECT
    SUM(price) AS total_sales,
    COUNT(*) AS total_sales_volume
FROM order_items;

-- 4. 统计销售额最高的前十个商品
SELECT
    product_id,
    SUM(price) AS sales_amount
FROM order_items
GROUP BY product_id
ORDER BY sales_amount DESC
LIMIT 10;

-- 5. 统计不同支付方式的金额及金额占比
SELECT
    payment_type,
    SUM(payment_value) AS total_payment,
    ROUND(
        SUM(payment_value) * 100.0 /
        (SELECT SUM(payment_value) FROM order_payments),
        2
    ) AS payment_percentage
FROM order_payments
GROUP BY payment_type
ORDER BY total_payment DESC;
"""

with open(sql_file, "a", encoding="utf-8") as file:
    file.write("\n" + summary_content)

print("SQL 综合分析语句写入完成")
````
