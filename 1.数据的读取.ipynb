```python
 #day1：数据读取、表结构、主外键，目标，理解数据结构


#导入库
import pandas as pd
import os
import glob

from pandas.core.interchange import dataframe
```

```python
#读取数据
data_path = './data/'
```

```python
#读取所有数据
csv_files =glob.glob(data_path + '*.csv')
```

```python
#建立字典
dataframes={}
```

```python
#遍历数据
for file in csv_files:
    name = os.path.basename(file).replace('.csv','')#替换将后缀替换为空
    dataframes[name] = pd.read_csv(file)
```

```python
import os

print("当前工作目录：", os.getcwd())
print("数据目录是否存在：", os.path.exists("./data/"))
```

```python
#统计表数量
print('成功读取到的表数量:',len(dataframes))
for name,df in dataframes.items():
    print(name,df.shape)
```

```python
#任务2
for name ,df in dataframes.items():
    print(f'\n====={name}=====')
    print('数据类型')
    print(df.dtypes)
```
